# Day 2 - Bronze / Silver / Gold Pipeline Skeleton (Insurance Lane)

**Draft only.** Deliberately small and explainable. Nothing here writes to production.

## Real column names vs. placeholders

Verified real columns in `source_extract/insurance_dataset.csv` (7 columns, 13,000 data rows):

`Age`, `Gender`, `Income`, `Marital_Status`, `Education`, `Occupation`, `Claim_Amount`

**PLACEHOLDER - these do NOT exist in this extract:**

- `PLACEHOLDER_claim_id` - the extract has **no claim id and no unique key of any kind**. Row identity is positional only.
- `PLACEHOLDER_claim_date` - the extract has **no date column**, so no incremental / CDC watermark is possible yet.

Treat both as open questions for the data owner, not as fields. Any uniqueness test or
incremental load in the run log is blocked until they are resolved.

## Layers

| Layer  | What it holds | Rule |
| ------ | ------------- | ---- |
| Bronze | Raw CSV read as text, plus load audit columns | Never edit a source value |
| Silver | Typed and trimmed, one row per bronze row | Tests gate promotion out of silver |
| Gold   | One row per `Occupation` for the claims manager | Built only if both tests pass |

## The two tests

1. **Schema test** - the expected 7 columns are present, in the expected order, with expected types.
2. **Null test** - the decision-critical columns `Claim_Amount` and `Age` contain no nulls.

```python
import pandas as pd
from pathlib import Path

evidence_dir = Path.home() / "Desktop" / "Persistent_Folder" / "Workspace" / "day-02-evidence"
evidence_data = evidence_dir / "source_extract" / "insurance_dataset.csv"

print("ready")
print(evidence_dir)
print(evidence_data)
print("exists:", evidence_data.exists())
```

## 1. Bronze - land the raw extract unchanged

Everything is read as text so a bad value cannot be silently coerced at load time.
Three audit columns are added. They are ours, not the source's, so they are prefixed `_`.

```python
def build_bronze(csv_path: Path) -> pd.DataFrame:
    """Land the raw CSV as text. No cleaning, no casting, no row drops."""
    bronze = pd.read_csv(csv_path, dtype=str, keep_default_na=False, na_values=[""])
    bronze["_source_file"] = csv_path.name
    bronze["_ingested_at_utc"] = pd.Timestamp.utcnow().isoformat()
    bronze["_bronze_row_num"] = range(1, len(bronze) + 1)  # positional stand-in for the missing claim id
    return bronze


bronze_claims = build_bronze(evidence_data)
print("bronze rows:", len(bronze_claims))
print("bronze columns:", list(bronze_claims.columns))
bronze_claims.head(3)
```

## 2. Silver - type and trim, one row per bronze row

`errors="coerce"` turns an unparseable number into a null instead of raising. That is on purpose:
we want the **null test below** to catch it and report how many, rather than the load dying with
a stack trace that tells the claims manager nothing.

Row count is asserted to be unchanged, so silver cannot quietly lose or duplicate rows.

```python
NUMERIC_COLUMNS = ["Age", "Income", "Claim_Amount"]
TEXT_COLUMNS = ["Gender", "Marital_Status", "Education", "Occupation"]


def build_silver(bronze: pd.DataFrame) -> pd.DataFrame:
    """Cast declared numerics, trim declared text. Keep one row per bronze row."""
    silver = bronze.copy()
    for col in NUMERIC_COLUMNS:
        silver[col] = pd.to_numeric(silver[col], errors="coerce")
    for col in TEXT_COLUMNS:
        silver[col] = silver[col].str.strip()
    assert len(silver) == len(bronze), "silver must not add or drop rows"
    return silver


silver_claims = build_silver(bronze_claims)
print("silver rows:", len(silver_claims))
print(silver_claims[NUMERIC_COLUMNS].dtypes.to_string())
silver_claims.head(3)
```

## 3. The two tests

Each test returns a small result dict rather than raising, so a failure becomes **evidence you can
paste into the run log** instead of a traceback.

- **Test 1, schema:** expected columns, expected order, expected type family.
- **Test 2, null:** zero nulls in `Claim_Amount` and `Age`.

Note the gap: there is no uniqueness test, because `PLACEHOLDER_claim_id` does not exist.

```python
# The contract. "number" means int or float; "text" means string.
EXPECTED_SCHEMA = {
    "Age": "number",
    "Gender": "text",
    "Income": "number",
    "Marital_Status": "text",
    "Education": "text",
    "Occupation": "text",
    "Claim_Amount": "number",
}
NOT_NULL_COLUMNS = ["Claim_Amount", "Age"]


def test_schema(df: pd.DataFrame) -> dict:
    """TEST 1 (schema): expected columns present, in order, with expected type family."""
    expected = list(EXPECTED_SCHEMA)
    actual = [c for c in df.columns if not c.startswith("_")]  # ignore our audit columns

    missing = [c for c in expected if c not in actual]
    unexpected = [c for c in actual if c not in expected]
    order_ok = actual == expected

    wrong_type = []
    for col, want in EXPECTED_SCHEMA.items():
        if col in missing:
            continue
        is_number = pd.api.types.is_numeric_dtype(df[col])
        got = "number" if is_number else "text"
        if got != want:
            wrong_type.append({"column": col, "expected": want, "actual": got})

    return {
        "test": "schema",
        "passed": not missing and not unexpected and order_ok and not wrong_type,
        "missing_columns": missing,
        "unexpected_columns": unexpected,
        "order_ok": order_ok,
        "wrong_type": wrong_type,
    }


def test_not_null(df: pd.DataFrame) -> dict:
    """TEST 2 (null): no nulls in the decision-critical columns."""
    counts = {}
    for col in NOT_NULL_COLUMNS:
        counts[col] = int(df[col].isna().sum()) if col in df.columns else "COLUMN MISSING"
    return {
        "test": "not_null",
        "passed": all(v == 0 for v in counts.values()),
        "null_counts": counts,
        "rows_checked": len(df),
    }


def run_tests(df: pd.DataFrame) -> tuple[bool, list[dict]]:
    results = [test_schema(df), test_not_null(df)]
    for r in results:
        print(("PASS  " if r["passed"] else "FAIL  ") + r["test"], "|", r)
    return all(r["passed"] for r in results), results


tests_passed, test_results = run_tests(silver_claims)
print()
print("all tests passed:", tests_passed)
```

## 4. Gold - build only if the tests passed

Grain: **one row per `Occupation`**. This is the gate the run log describes: aggregate only after
tests pass. If a test fails, gold is not built and the previous gold output stays in place.

```python
def build_gold(silver: pd.DataFrame) -> pd.DataFrame:
    """One row per Occupation. Claims summary for the claims operations manager."""
    gold = (
        silver.groupby("Occupation", dropna=False)
        .agg(
            claim_row_count=("Claim_Amount", "size"),
            total_claim_amount=("Claim_Amount", "sum"),
            avg_claim_amount=("Claim_Amount", "mean"),
            max_claim_amount=("Claim_Amount", "max"),
            avg_income=("Income", "mean"),
        )
        .round(2)
        .reset_index()
        .sort_values("total_claim_amount", ascending=False)
    )
    return gold


if tests_passed:
    gold_claims = build_gold(silver_claims)
    print("gold rows (one per Occupation):", len(gold_claims))
    display(gold_claims)
else:
    gold_claims = None
    print("GOLD NOT BUILT - a test failed. Diagnose, propose a fix, get approval, then rerun.")
```

## 5. Force a failure on purpose (negative control)

The clean extract has **zero nulls**, so both tests pass on the first run. A test that has never
failed is not yet evidence that it works. This cell breaks a **copy** of silver - the real
`silver_claims` is untouched - to produce the failed-test evidence Day 2 asks for.

It simulates two things at once: a renamed column (schema drift) and blanked amounts (quality drift).

```python
broken = silver_claims.copy()
broken = broken.rename(columns={"Claim_Amount": "ClaimAmount"})  # schema drift: renamed column
broken.loc[broken.index[:25], "Age"] = None                      # quality drift: 25 null ages

print("--- expecting BOTH tests to FAIL ---")
broken_passed, broken_results = run_tests(broken)
print()
print("all tests passed:", broken_passed, "<- must be False for the negative control to be valid")
```

## 7. One risk if the output is wrong

**The tests can pass while gold is still wrong, because there is no `PLACEHOLDER_claim_id` to test
uniqueness against.** If the extract ever contains duplicated claim rows, both tests still pass -
the schema is intact and no nulls appear - but `total_claim_amount` per `Occupation` is inflated by
the duplicates. The claims manager would approve a rerun on green tests and publish overstated
claim totals. Mitigation: get a real claim id from the data owner and add a uniqueness test before
this skeleton is trusted for a decision.
